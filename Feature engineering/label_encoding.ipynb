Label Encoding 

Label encoding and ordinal encoding are two techniques used to encode categorical data as numericla data.

Label encoding involves assigning a unique numerical label to each categories in the variable. The labels are usually assingned in alphabetical order or based on the frequency of the categories. For example, if we have a categorical variable "color" with three possible values (red, green ,blue) we can represent it using label encoding as follows:

1. Red 1
2. Green 2
3. Blue 3

```python
import pandas as pd 
from sklearn.preprocessing import LabelEncoder
import seaborn as sns 
```

```python
## Create a simple dataframe 

df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'blue', 'red']
})
```

```python
df.head()
```

```python
lbl_encoder = LabelEncoder()
```

```python
encoded = lbl_encoder.fit_transform(df[['color']])
print(encoded)
```

```python
## for adding  new values 
lbl_encoder.transform([['blue']])
```

#### Ordinal encoding 
when we have to decide the rank or order. In this technique, each category is assigned a numerical value based on its positioiin in the order. Example if we have a cstegorical variable " education level" with four possible values 

1. High school: 1
2. College: 2
3. Graduate: 3
4. Post-Grauduate: 4

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.DataFrame({
    'size': ['small', 'medium', 'large','small', 'medium', 'large', 'small', 'large']
})
```

```python
df
```

```python
## assigning ranking 
encoder = OrdinalEncoder(categories= [['small', 'medium', 'large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
print(encoder.transform([['small']]))
```
