```python
import numpy as np 
```

```python
lst_marks = [45,58,74,85,96,70,48,59,12,35,45,78,96,36,45,25,74]
minimum, Q1, median, Q3, maximum =np.quantile(lst_marks, [0,0.25,0.50,.75,1.0])
```

```python
IQR = Q3-Q1
print(IQR)
```

```python
lower_fence = Q1 -1.5*(IQR)
higher_fence = Q3 + 1.5*(IQR)


print(lower_fence)
print(higher_fence)
```
