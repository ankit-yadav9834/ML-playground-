```python
import pandas as pd 
from sklearn.preprocessing import OneHotEncoder
import seaborn as sns 
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
encoder = OneHotEncoder(sparse_output=False)
```

```python
encoded = encoder.fit_transform(tips[['sex']])
```

```python
encoded_df = pd.DataFrame(
    encoded,
    columns= encoder.get_feature_names_out(['sex'])
)
```

```python
encoded_df
```

```python
categories = ['sex', 'smoker', 'day', 'time']
```

```python
encoder = OneHotEncoder(sparse_output= False)
```

```python
encoded = encoder.fit_transform(tips[categories])
```

```python
encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(categories)
)
```
