### Nominal/OHE encoding 

One hot encoding also known as nomial encoding is a technique used to represent categorical data as numerical data, which is more suitable for machine learnipng algorithms, in this technique, each category is represented as a binary vector where each bit correspponds to a unique category. For example, if we have a categorical variable "color" wiyh threee possible values (red, greeen , blue) we can represent it on hotcoding encoding as follows:

1. Red: [1,0,0]
2. Green [0,1,0]
3. Blue[0,0,1]


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
## Create a simple dataframe 

df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'blue', 'red']
})
```

```python
df.head()
```

```python
## Create an instance of oneHotEncoder

encoder = OneHotEncoder()
```

```python
# Fit and Transform 
encoded = encoder.fit_transform(df[['color']]).toarray() ## alphabetically sorting is done in blue B is  then so one 
```

```python
encoder_df = pd.DataFrame(encoded, columns= encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
pd.concat([df, encoder_df], axis=1)
```
