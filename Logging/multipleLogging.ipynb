Logging with multiple loggers 

You can create multiple loggers for different parts of your application

```python
import logging 

#create a logger for module1

logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)

## create a logger for module2

logger2 = logging.getLogger("module2")
logger2.setLevel(logging.WARNING)

# creating logging setting 
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s -%(levelname)s-%(message)s',
    datefmt= '%y-%m-%d  %H:%M:%S'
)
```

```python
## log messages for different messages 
logger1.debug("This is debug message for module2")
logger2.warning("This is a warning message for module2")
logger2.error("This is error messages ")
```

```python
import logging 

# creating logging setting FIRST
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%y-%m-%d %H:%M:%S'
)

# create a logger for module1
logger1 = logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)

# create a logger for module2
logger2 = logging.getLogger("module2")
logger2.setLevel(logging.WARNING)

# log messages
logger1.debug("This is debug message for module1")
logger2.warning("This is a warning message for module2")
logger2.error("This is error message")
```

```python
import logging

logging.shutdown()

for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)
```

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%y-%m-%d %H:%M:%S',
    force=True   # IMPORTANT
)

logger1 = logging.getLogger("module1")
logger2 = logging.getLogger("module2")

logger1.debug("This is debug message for module1")
logger2.warning("This is a warning message for module2")
logger2.error("This is error message")
```
