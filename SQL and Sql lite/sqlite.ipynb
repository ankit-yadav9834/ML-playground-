SQL (Structured Query Language) is a standard language for managing realtional database.
SOLite is a self contained, serverless and zero configuration database engine that is widely used for embedded database system.

```python
import sqlite3
```

```python
# Connecting to an SQlite database

connection = sqlite3.connect('example.db')
connection
```

```python
#Cursor isliye use hota hai kyunki SQLite database ko instructions cursor ke through hi milti hain.
cursor = connection.cursor()
```

```python
# creating a table


cursor.execute(''' 
Create Table If Not Exists employee(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')

# commit the changes 
connection.commit()
```

```python

cursor.execute(''' 
Select * from employee
    
''')
```

```python
# insert the data in sqlite3

cursor.execute("""
INSERT INTO employee (name, age, department)
VALUES ('Krish', 32, 'Data Scientist')
""")

cursor.execute("""
INSERT INTO employee (name, age, department)
VALUES ('Ankit', 25, 'Software Engineer')
""")

cursor.execute("""
INSERT INTO employee (name, age, department)
VALUES ('Deepika', 24, 'Data Analyst')
""")

connection.commit()
```

```python
# Query the data from the table


cursor.execute('Select * from employee')
rows = cursor.fetchall()

#print the data 

for row in rows:
    print(row)
```

```python
# Update the data into table 

cursor.execute('''
UPDATE employee
Set age = 85 
where name = "Krish"        
''')

connection.commit()

# printing 
cursor.execute('Select * from employee')
rows = cursor.fetchall()

#print the data 

for row in rows:
    print(row)
```

```python
# Delete the data from the table 

cursor.execute('''
Delete from employee
               where name = "Krish"
''')

# printing 
cursor.execute('Select * from employee')
rows = cursor.fetchall()

#print the data 

for row in rows:
    print(row)
```

```python
## close thw connectionj

cursor.close()
```
