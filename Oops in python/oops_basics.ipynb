```python
## Instance variabled and methods 
class dog:
    ## construcutor 
    def __init__(self, name,age):   # self -> Ye method ko batata hai ki wo kis object ke data (attributes) ke saath kaam kar raha hai
        self.name = name # yha self name ke sath kam kr raha hai 
        self.age = age   # yhaa self age ke data ke sath kam kr raha hai 

## create objects 
dog1 = dog("kik", 3)
print(dog1) ## this print the address 

## to retrive the variable
print(dog1.age)
print(dog1.name)

```

```python
## define a class with instance methods

class dog:
    def __init__(self, name, age):
        self. name = name
        self.age = age 

    def bark(self):
        print(f"{self.name} says woof")


dog1 = dog("buddy",4)
dog1.bark()

        
```

```python
## modeling a bank account 

## define a class for a bank ac
class BankAccount:
    def __init__(self, owner, balance = 0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} is deposited and new balance is {self.balance}")

    def withdraw(self, amount):
        if amount> self.balance:
            print("Insufficient balance")
        else:
            self.balance-= amount
            print(f"{amount} is withdrwan and avaiblable balance is {self.balance}")

    def get_balance(self):
        print(f"your current balance is {self.balance}")


## to retrive the information of owner
account = BankAccount("ABC", 500)
print(account.owner)
account.deposit(50000)
account.withdraw (40)
account.get_balance()

```
