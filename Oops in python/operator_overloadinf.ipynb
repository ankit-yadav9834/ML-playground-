Operator overloading


Operator overloading giving operators (+ - * == >) a custom meaning for your class.

Python lets objects decide what + or == means.

Why it exists (real reason)

Default operators only work for:

numbers

strings

lists

```python
'''
Comparison

__eq__ → checks equality (==)

__ne__ → not equal (!=)

__lt__ → less than (<)

__gt__ → greater than (>)

__le__ → less than or equal (<=)

__ge__ → greater than or equal (>=
'''
```

```python
# mathematical operation 
class Vector:
    def __init__(self,x,y):
        self.x= x
        self.y=y

    def __add__(self,other):
        return Vector(self.x +other.x, self.y+ other.y)

    def __sub__(self,other):
        return Vector(self.x -other.x, self.y- other.y)

    def __mul__(self,other):
        return Vector(self.x *other.x, self.y* other.y)

    def __eq__(self,other):
        return Vector(self.x ==other.x, self.y== other.y)

    def __repr__(self):
        return f"vector ({self.x}, {self.y})"

    
    ## create objects of the vector class

    v1 = Vector(10,10)
    v2= Vector(100,100)

    print(v1+v2)
    print(v1-v2)
    print(v1*v2)
    
```
