#   Magic Methods

Magic methods in python, also knowns as dunder methods( double underscore methods) are special methods that start and end with double underscores. These methods enable you to define the behaviour of the objects for built-in operations such as arithmetic, comparision and more.

Magic merhods are predefined methods thar you can override to change the behaviour of yoiur objects. Some common magic methods are :



```python
'''
__init: Initializes a new instsance of a class
__str: Return a string representation of an object
__repr: Return an official string representaiton of an object
__len: Return the length of an object
__getitem: Gets an item from an container
__setitems: Sets an item ina container.
 '''
```

```python
# magic methods

class Person:
    def __init__(self, name,age):
        self.name = name
        self.age = age 

        # str methods , we have overriden this 
    def __str__(self):
        return f"{self.name},{self.age} years old"
    
    def __repr__(self):
        return f"person (name= {self.name})"
    
    
person = Person("ABCD", 25)
print(person)
print(repr(person))
```
