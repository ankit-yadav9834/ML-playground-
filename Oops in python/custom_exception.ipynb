### Custom exception (Raise and Throw an exception)

```python
class Error(Exception):
    pass

class dobException(Error):
    pass

 
```

```python
year = int(input("Enter the DOB : "))
age = 2026 -year

try:

    if age<=30 and age >=20:
        print("Your age is eligible")

    else:
        raise dobException
    
except dobException:
    print("sorry you are not eligible")


```
