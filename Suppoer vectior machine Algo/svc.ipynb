```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification

X,y = make_classification(n_samples=1000, n_classes=2, n_features=2, n_clusters_per_class= 1, n_redundant=0)
```

```python
X
```

```python
y
```

```python
pd.DataFrame(X)[0]
```

```python
sns.scatterplot(x=pd.DataFrame(X)[0], y=pd.DataFrame(X)[1], hue=y)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test,y_train,y_test = train_test_split(X,y, test_size=0.25, random_state=10)
```

```python
from sklearn.svm import SVC
svc = SVC(kernel= 'linear')
```

```python
svc.fit(X_train, y_train)
```

```python
# prediction 
y_pred= svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
rbf = SVC(kernel= 'rbf')
```

```python
rbf.fit(X_train,y_train)
```

```python
y1_pred= rbf.predict(X_test)
```

```python
print(classification_report(y_test, y1_pred))
```

```python
print(confusion_matrix(y_test, y1_pred))
```

```python
polynomial = SVC(kernel= 'poly')
```

```python
polynomial.fit(X_train,y_train)
```

```python
y2_pred=polynomial.predict(X_test)
```

```python
print(classification_report(y_test, y2_pred))
```

```python
print(confusion_matrix(y_test, y2_pred))
```

## HyperParameter with SVC

```python
from sklearn.model_selection import GridSearchCV

param= {
    'C':[100,10,1,0.1,0.01],
    'gamma':[1,0.1,0.01,0.001],
    'kernel': ['rbf']
}
```

```python
grid= GridSearchCV(SVC(), param_grid= param,cv=5, verbose=3)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y4_pred=grid.predict(X_test)
print(classification_report(y_test, y4_pred))
print(confusion_matrix(y_test, y4_pred))
```
