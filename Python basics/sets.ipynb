## Sets


```python
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(type(my_empty_set))
```

```python
my_empty_set = set([1,2,3,4,5,6,7])
print(type(my_empty_set))
print(my_empty_set)
```

```python
my_empty_set = set([1,2,3,4,4,4,5,6,7])
print(type(my_empty_set))
print(my_empty_set)
```

```python
## to add & remove the elements in a set 

my_set.add(8)
print(my_set)

my_set.remove(2)
print(my_set)
```

```python
my_set.discard(10)
print(my_set)
```

```python
## pop method 

removed_elemtns = my_set.pop()
print(removed_elemtns)
print(my_set)
```

## set membership test

```python
my_set = (1,2,3,4,5)
print(3 in my_set)
print(10 in my_set)
```

## mathematical operations 

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,9,10,11}

## union 
union_set = set1.union(set2)
print(union_set)

## Intersections 
intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
## Difference jo set 1 me common element hoge vo print kar dega

set1 = {1,2,3,4,5,6}
set2 = {4,5,9,10,11}

print(set1.difference(set2)
)

print(set2.difference(set1))
```

```python
## symentric differece mtlb dono me jo common element h unhe remove kr ke 
## baaki ke element ko print krna 

set1.symmetric_difference(set2)
```

## sets methods 

```python
set1= {1,2,3,4,5,6}
set2= {4,5,6,}

## is subset 
print(set1.issubset(set2))

## is superset 

print(set1.issuperset(set2))

```

```python
## to remove duplicate elements from list : u convert it into set and rthge 
lst = [1,2,3,4,5,4,4]
set(lst)
```

```python
### counting unique words in text 

text = " In this tutorial we are discussing about sets "

words = text.split()
#print(words)

## covert list of words to set to get unique words 

uniques_words = set(words)
print(uniques_words)
print(len(uniques_words))
print(len(text))
```
