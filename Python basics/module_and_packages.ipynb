In python modules and packages help organize and reuse code. Here's comprehensive guide on how to import them.



```python
import math
math.sqrt(100)
```

```python
## data serialization
## converting into JSON 

import json

data = {'name': 'Ankit', 'age':1 } ## this is my dictionary, now if I want to convert this into JSON then,
json_str= json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data =json.loads(json_str) ## this we use when we again want to convert into dictionary 
print(parsed_data)
print(type(parsed_data))

```

```python
## CSV file

import csv

with open('example.csv', mode= 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Ankit', 21])


with open('example.csv', mode= 'r') as file:
    reader= csv.reader (file)
    for row in reader:
        print(row)
```

```python
# regular expression 

import re

pattern = '\d+'          '''- \d → special regex token meaning “any digit ( 0to 9)
                         - + → quantifier meaning “one or more of the preceding thing”.
                        - So \d+ matches a sequence of digits.
                        '''
text = 'there are 123 apples'
match = re.search(pattern,text)
print(match.group())
```

```python
#If you want to match letters instead of digits: 

import re 

pattern2= '[a-zA-Z]+'
text2 = 'there are 123 apples in a cart '
match2 = re.search(pattern2,text2)
print(match2.group()) 
```
