#  Dictionaries 

Dictionaries are the unordered collections of items. They store data in key -value pairs. key must be unique and immutable (eg :- stringd, numbers, or tuples) while values can be of any types.

```python
## Creating dictionaries 

empty_dict ={}
print(type(empty_dict))
```

```python
student = {"name": "Ankit", "age": 20}
print(student)

## Single key is always used:- as below example i have created 
#two key "name" but in the output the key "name " have "value" 24 
# bcoz the recent one is considered 


student = {"name": "Ankit", "age": 20, 'name': 24}
print(student)

## correct one only 1 key value 
student = {"name": "Ankit", "age": 20}
print(student)


```

```python
## Accessing dictionary elements 

print(student['name'])
print(student['age'])


## accessing using get() function method 

print(student.get('name'))
print(student.get('last_name'))
print(student.get('last_name', "Not avaible"))
```

```python
## Modifying Dictionary elements 
# Dict are mutable, so we can change, add, remove or update elements

print(student)
```

```python
student["age"] = 33 # updated the key
print(student)
student ["address"]= "India"  # added a new key and value pair 
print(student)
```

```python
## delete the key

del student ['age'] # delete ket and value pair 
print(student)

```

# Dictionaries method


```python
keys = student.keys() # get all the keys
print(keys)

values = student.values() # get all the value
print(values)

items = student.items()  # get all the keys and values 
print(items)
```

## Shallow copy verry ImP

```python
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"]= "krish"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() # shallow copy 
print(student)
print(student_copy1)
```

```python
student ["name"]= "krish 3"
print(student_copy1)
print(student)
```

```python
## Iterating over 
##  You can iterate dictionaries ,keys, values 


## Iterating over 
for key in student.keys():
    print(key)

## iterate over values 

for value in student.values():
    print(value)


## iterate over the key values pairs

for key, value in student.items():
    print(f"{key}:{value}")
```

```python
## nested dictionaries 

students= {
    "student1": {"Name":"Ankit", "age":19},
    "student2": {"Name":"Saket", "age":18}

}
print(students)
```

```python
## Access nested dictionaries elements

print(students["student2"]["Name"]) 

print(students["student2"]["age"]) 
```

```python
students.items()
```

```python
## Iterating over nested dictionaries 

for student_id , student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key, value in student_info. items():
        print(f"{key}: {value}")
```

```python
## dictionary comprehenssion 

squares = {x : x**2 for x in range (5)}
print(squares)
```

```python
## conditional dictionary comprehenssion 

evens = {x : x**2 for x in range(10) if x% 2==0}
print(evens)
```

```python
## practical examples 
# use a dictionary to count the frequency of elements in list

numbers = [1,2,3,4,5,5,2,3,4,1]

frequncy ={}

for number in numbers :
    if number in frequncy:
        frequncy[number]+=1
    
    else:
        frequncy[number]=1

print(frequncy)
```

```python
## Merge two dictionaries 

dict1 = {"A":1, "B":2}
dict2 = {"B":3,"C":4}

merge_dict = {**dict1, **dict2}
print(merge_dict)
```
