## Exception handling in Python 

Exception handling in python allows you to handle errors gracefully and take corrective action without stopping the execution of the program.





What are Exceptions?

Exceptions are events that disrupt the normal flow of a program. They occur when an error is encountered during the program execution. Common exception includes:

ZeroDivisionErorr : Dividing by zero
FileNotFoundError : File not found 
ValueError : Invalid error
TypeError : Invalid type

```python
try:
    num = int(input("enter a number : "))
    result = 10/ num 

except ZeroDivisionError as e:
    print(e)
    print("enter the den >0")

except ValueError as ee:
    print(ee)
    print("enter the correct int value")

except Exception as ex1:
    print(ex1)

```

```python
# try except, else block we use else when we want to get the value of the result

try :
    num = int(input("entert a number :"))
    result = 100/num

except ZeroDivisionError as e:
    print(e)
    print("enter the den >0")

except ValueError as ee:
    print(ee)
    print("enter the correct int value")

except Exception as ex1:
    print(ex1)

else:
    print(f"the result is {result}")



```

```python
## try, except , else and finally 
## finally will executed when there is no error 

try :
    num = int(input("entert a number :"))
    result = 100/num

except ZeroDivisionError as e:
    print(e)
    print("enter the den >0")

except ValueError as ee:
    print(ee)
    print("enter the correct int value")

else:
    print(f"the result is {result}")

finally:
    print("execution completed")






```
