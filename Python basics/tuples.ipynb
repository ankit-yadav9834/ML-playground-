## Tuples
orders collections of items that are immutable.

```python
## creating tuple

empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
numbers = tuple([1,23,4,5,6])
numbers
```

```python
mixed_tuple = ("sfda",25 , 2.5, True)
mixed_tuple
```

```python
## accessing tuple elements 

print(numbers[0])
print(numbers[-1])
```

```python
## slicing of tuple reverse order 
numbers[::-1]
```

```python
## tuple operation concatination , multiplication

print(numbers + mixed_tuple)

print(mixed_tuple *3)
```

```python
## Immutable nature of tuples

lst = [1,2,3,4,5]
print(lst)

lst[1] = "jhdf"
print(lst)



numbers[1]= "bhfv"
```

```python
## tuple methods 

print(numbers.count(1))
print(numbers.index(1))
```

```python
## packing and unpacking tuple 

packed_tuple = 1,"jhf",30,3.14
print(packed_tuple)
```

```python
## how to unpack the tuple 
a,b,c,d = packed_tuple

print(a)
print(b)
print(c)
print(d)
```

```python
## unpacking with star (*) iska mtlb hia jaha se tart hota hai uske bad se 
# le kr last ke pahle tk ke elemenmts  

numbers = (1,2,3,4,5,6)
first, *middle , last = numbers

print(first)
print(middle)
print(last)
```

```python
##  nested tuples 

lst = [[1,2,3,4],[4,5,6],["a","b","c"]]
print(lst)
print(lst[0])
print(lst[0][2])
print(lst[0][0:3])
```

```python
### nested tuples 

nested_tuples = ((1,2,3,4),(4,5,6),("a","b","c"))
print(nested_tuples)
print(nested_tuples[0][2])
```

```python
## iterating over nested tupple 

for sub_tuple in nested_tuples:
    for items in sub_tuple:
        print(items,end =" ")
    print()
```
