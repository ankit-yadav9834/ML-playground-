
# list comprehension syntax

# basic syntax 
[expression for item in iterable ]

with conditional logic [expression for item in iterable if condition]



```python
## basic list comprehenssion 
sqr = [num**2 for num in range (10)]
print(sqr)
```

## REAL WORLD USE CASES OF LIST 

## Example 1. Manage A to do list 

create a To Do list to keep track of Tasks 

```python
to_do_list = ["Buys Vegies", "Pay Bills", "Clean my bike"]

# adding tasks
to_do_list.append("Go for a Run")
to_do_list.append("Attend lectures")

# removing completed tasks
to_do_list.remove("Buys Vegies")
to_do_list.remove("Pay Bills")

# reminder
if "Clean my bike" in to_do_list:
    print("Don't forget to do this!")

# print list
for task in to_do_list:
    print(f"- {task}")
```

## Oraganizing student grades 

Create a list to store and calculate average grades for students


```python
# student grades 
student_grades = [85,96,74,14,25,36]

# adding a new grades 
student_grades.append(94)

## finding the average grades 
average_grades = sum(student_grades)/len(student_grades)
print(f"Average grades of Students are :- {average_grades: .2f}")

##finding the highest and lowest grades
highest_grades = max(student_grades)
lowest_grades = min(student_grades)
print(f"highest grades are :- {highest_grades}")
print(f"lowest grades are :- {lowest_grades}")
 
```

## Example 3: Managing An Inventory 

Use a list to manage inventory items in a store

```python
# Initial inventory list
inventory = ["Milk", "Bread", "Eggs", "Rice", "Oil"]

# Adding new items to inventory
inventory.append("Sugar")
inventory.append("Biscuits")

# Selling items (removing from inventory)
inventory.remove("Milk")
inventory.remove("Eggs")

# Check if a specific item is available
item_to_check = "Rice"
if item_to_check in inventory:
    print(f"{item_to_check} is available in stock.")
else:
    print(f"{item_to_check} is not available in stock.")

# Display the updated inventory
print("\nCurrent Inventory Items:")
for item in inventory:
    print(f"- {item}")
    
```
