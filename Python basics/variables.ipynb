```python
 ## dtermine if a year is keap year using the nested loop

year = int (input("enter the year "))

if (year%4==0):
    if year%100 == 0:
        if year % 400==0:
            print(year,"this is a leap year")
        else:
            print(year,"this is not a leap year")
    else:
        print(year, "this is year ")

else:
    print(year,"this is not a leap year")
```

```python
 ## Loop control statement 
## break statement exits the looop permanently 


for i in range (100):
    if i ==10:
        break
    print(i)
```

```python
## continue statement 
# the continue statement skips the current iteration and continues with the next 

for i in range (10):
    if i%2 ==0:
        continue
    print(i)
```

```python
## pass 
##the pass statement is the null operation and does nothing


for i in range (5):
    if i == 3:
        pass
    print(i)
```
