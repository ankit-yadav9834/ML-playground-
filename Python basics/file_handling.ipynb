## File operation : Read and write Files 

File handling is a crusial part of any progrraming language. Python provides built in functions and methods to read from and write to files, both text and binary. This will cover the basic of file handling ,icluding reading and writing text files and binary files.
 

```python
## read a whole file

with open('example.txt', 'r') as file :
    content = file.read()
    print(content)
```

```python
## read line by line 

with open('example.txt', 'r') as file:
    for line in file:
        print(line)
```

```python
## to remove the space means new line space as u seen in above o/p we use .strip()

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

```python
## writing a file(overwriting)

with open('example.txt', 'w') as file:
    file.write("Hellow World ! \n")
    file.write("this is a new line\n")
```

```python
## write a file (without overwriting)

with open('example.txt', 'a') as file:
    file.write("Append mode is taking place! \n")
```

```python
## writing a list of lines to a file

lines = ['first line\n ','Second line\n', 'third line\n']
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
## binary file and having extension .bin

data = b'\x00\x01x\x02\x03\x04\x05'
with open('example.bin', 'wb') as file: # wb means write byte 
    file.write(data)
```

```python
## binary file and having extension .bin

data = b'\x00\x01x\x02\x03\x04\x05'
with open('example.bin', 'rb') as file: # wb means write byte 
    content = file.read()
    print(content)
```

```python
## practical examples 
##  read the content from source text file and write to a destination text file

with open ('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
## read a text file and count the number of lines, words and characters. 
#Counting lines, words and character in a text file


with open('example.txt', 'r') as count_file:

    lines_count = 0
    words_count = 0
    char_count = 0

    for line in count_file:
        lines_count+= 1
        words = line.split()
        words_count +=len(words)
        char_count+= len(line)


print("total lines:- ", lines_count)
print("total words:- ", words_count)
print("total characters:-", char_count)
```

```python
# to check if the file is present or not 

path = 'example1.txt'
if os.path.exists(path):
    print(f"the path '{path} exists")

else:
    print(f"the path'{path} does not exists")
```

```python
# checking if a path is a file or directory 

import os

path = 'example1.txt' 
if os.path.isfile(path):
    print(f"the path '{path} is a file")
elif os.path.isdir(path):
    print(f"this path '{path} is a directory")
else:

    print(f"this path '{path} is not a file or directory")
```

```python
## getting the absolute path 

relative_path = 'example.txt'

absolute_path= os.path.abspath(relative_path)
print(absolute_path)
```
