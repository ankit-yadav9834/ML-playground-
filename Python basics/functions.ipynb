```python
#  variable length arguments
##      Positional and keywords arguments (when we )

def print_numbers(*args):
    for number in args :
        print(number)
```

```python
print_numbers(1,2,3,4,5,5,67,8, "krish")
```

```python
## keywords arguments (all the parameters are in the form of key value pairs)

def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
        
```

```python
print_details(name= "Ankit", age= 21)
```

```python
def print_details(*args,**kwargs):
    for val in args :
        print(f"{val}")

    
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_details(1,2,3,4,5,6,"krish",name= "Ankit", age= 21)
```

## Functions example

```python
## EX 1: convert temperature 

def convert_temperature(temp, unit):
    """this function convert temperature b/w celsius and fahrenhiet"""

    if unit=="c":
        return temp *9/5 + 32 ## converting celsius to fahrenhiet 
    
    elif unit == "f":
        return (temp-32)*5/9 ## give fah-> celsius
    

    else:
        return None

print(convert_temperature(25,'c'))
print(convert_temperature(25,'f'))



```

```python
## Ex 2: pass strength checker

def is_strong_password(password):
    if len(password)<8:
        return False
    
    if not any(char.isdigit() for char in password ):
        return False
    
    if not any(char.islower() for char in password):
        return False
    
    if not any(char.isupper() for char in password):
        return False
    
    if not any( char in '!@#$%^&*()_+' for char in password):
        return False
    
    return True

## calling this function 

print(is_strong_password("WeakPwd"))
print(is_strong_password("Str@ong1232"))
```

```python
## Ex 3 : Check if palindrome 

def check_palindrome(s):
    s = s.lower().replace(" "," ")
    return s==s[::-1]

print(check_palindrome("A man a plan a canal Panama"))



    
```
