```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
import pandas as pd
%matplotlib inline 
```

```python
# load the datset 
from sklearn.datasets import load_breast_cancer
```

```python
cancer_datset = load_breast_cancer()
```

```python
cancer_datset.keys()
```

```python
print(cancer_datset.DESCR)
```

```python
df = pd.DataFrame(cancer_datset['data'],columns= cancer_datset['feature_names'])
```

```python
df
```

```python
# standarization 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
```

```python
scaled_data
```

```python
# applying pca algorithm 
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
data_pca = pca.fit_transform(scaled_data)
```

```python
data_pca
```

```python
pca.explained_variance_
```

```python
# Visualization 
plt.Figure(figsize=(8,6))
plt.scatter(data_pca[:,0], data_pca[:,1], c= cancer_datset['target'], cmap= 'plasma')
plt.xlabel('first principal component')
plt.ylabel('first principal component')
```
