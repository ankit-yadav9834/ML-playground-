#### Iterators

Iterators are advanced python that allows for effcient looping and memory management. Iterators provide a way to access elements of a collection sequantially without exposing the underlying structure 

```python
my_list = [1,2,3,4,5,6]

for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
print(my_list)
```

```python
## How we can cretae Iterator 

iterator = iter(my_list)
print(type(iterator))
```

```python
iterator 
```

```python
# Iterate through all the element 
# har bar code run krne se ye 1 se 2 se 3 se4 and lst 5 tk aayega lazy loading 
next(iterator)
```

```python
# Exception hadndling 
try:
    print(next(iterator))

except StopIteration:
    print("There are no elements in the iterator ")
```

WHEN to use Iterator

Jab data bahut bada ho (files, logs, DB records)

Jab poora data ek saath memory me nahi chahiye

Jab one-time sequential processing karni ho

Jab infinite ya very large sequence ho (e.g. range, streams)

Jab real-time data aa raha ho (API, sensors, network)

Jab for-loop ke liye sirf read-only traversal chahiye

WHY to use Iterator

Memory efficient – ek element at a time

Fast start – bina poori collection banaye

Lazy evaluation – data tab generate hota hai jab chahiye

Scales better – big data friendly

Cleaner looping – for loop naturally support karta hai

Avoids unnecessary storage
