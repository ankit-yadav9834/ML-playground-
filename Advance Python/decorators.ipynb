### DECORATORS

Decorators are a powerful and flexible feature in python that allows you to modify the behaviour of a function or class method. THey are commonly used to add functionality to functionsor methods without modifyinh their actual code.

```python
    ## function copy 
## closures 
## decorator
```

```python
## function copy 

def welcome():
    return "Welcome to INDIA !"

welcome()
```

```python
wel = welcome
print(wel())   # this is function copy 


del welcome # delete the welcome funciton
print(wel())
```

```python
# closures (function inside function)

def main_welcome(msg):
    

    def sub_welcone_method():
        print("welcome to this playgraround")
        print(msg) # acess the msg from uter function

        print("please learn something ")

    return sub_welcone_method()
```

```python
main_welcome("Welcome everyone")
```

```python
# Decorator
def main_welcome(func):
    

    def sub_welcone_method():
        print("welcome to this playgraround")
        func()

        print("please learn something ")

    return sub_welcone_method()
```

```python
def cource_introduction():
    print("this is an advance python cource")

cource_introduction()
```

```python
main_welcome(cource_introduction)
```

```python
@main_welcome # @ is the keyword for decorator 
def cource_introduction():
    print("this is an advance python cource")
```

```python
# 2nd example

def my_decorator(func):

    def wrapper():

        print("Something is happening before the function is called")
        
        func()
        
        print("Something is happening after the fucntion is called")

    return wrapper
```

```python
@my_decorator

def say_hello():
    print("Hlw")
```

```python
say_hello()
```
