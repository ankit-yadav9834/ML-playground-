### Generators 

Generators are a single way to create iterators. They use the yield keyword to produce a series of a vlaues lazily, which means they generate values on the fly and do no store them in memory


```python
def sqaure(n):
    for i in range(3):
        yield i**2 # we use yield keyword in genrator 
```

```python
for i in sqaure(3):
    print(i)
```

```python
a = sqaure(3)
next(a)
```

```python
next(a)
```

```python
def my_genrator():
    yield 1 
    yield 2
    yield 3
    
```

```python
gen = my_genrator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

#### Practical example : Reading large files

Genrators are particularly useful for reading large files becasue they allow you to process one line at a time without loading the enitre file into memory.

```python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
import os 
print(os.getcwd())
```

```python
file_path = 'large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```
