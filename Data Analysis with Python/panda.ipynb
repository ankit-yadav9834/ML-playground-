##### Pandas- DataFrame and Series

Pandas is a powerful data manipulation liberary in python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A series is one dimensional array like object, while a DataFrame is a 2D, size- mutable and potentially hetrogeneous tabular data structure with labelled axes (rows & columns)

```python
import pandas as pd
```

A Pandas Series is a 1D array like object that can hold any data  type. It is similar to a column in a table


```python
# Series 

data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
```

```python
## creating series from dictinary elements
data= {'a':1,'b':2,'c':3}
series_dict = pd.Series(data)
print(series_dict) # here a b c are the index and 1 2 3 are the vlaues 
```

```python
data = [1,3,45]
index = ['a','b','c']
pd.Series(data, index= index)
```

```python
# DataFrame 
## create a DataFrame from a dictionary of list

data = {
    'Name': ['Alice', 'John', 'Charle'],
    'Age': [25,52,45],
    'City': ['Banglore', 'NewYork', 'Chicago']
}
df= pd.DataFrame(data)
print(df)
```

```python
import numpy as np
np.array(df)
```
