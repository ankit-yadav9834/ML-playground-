```python
import pandas as pd
```

```python
# Handling missing values 

df.isnull().any(axis=0)
```

```python
df.isnull().any(axis=1)
```

```python
df.isnull().sum() # this tells kitni values missing hai jaise ost_id me koi missing valuue hai
```

```python
# agar missing values hai aur hame uski place fill krni hai then we use inbuilt function fillna()

df.fillna(0) # 0 is the by default value 
```

```python
# filling missing value with the mean of the column 

df['likes_fillNA']= df['likes'].fillna(df['likes'].mean())
df

# This code means if there is missing values in the dataset then we made 
# new row name likes_fillNA which include the mean of the likes, so when 
# there is missing values it will fill with the mean of the likes.
```

```python
df.dtypes
```

```python
# Renaming column
df= df.rename(columns={'lang':'Language'})
df.head()
```

```python
# Changing DataTypes

df['Value_new']= df['likes'].astype(float)
df.head()
```

```python
# applying funciton to one particular function by usingf apply(function)

df['Dislikes'] = df['likes'].apply(lambda x : x*2)
df.head()
```

```python
# Data Aggregatingand Grouping

grouped_mean = df.groupby('post_id')['likes'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['stance', 'timestamp'])['likes'].sum() # same for mean instaead of sum() place mean()
print(grouped_sum)
```

```python
## aggregate mulltiple function

grouped_agg = df.groupby('likes')['comments'].agg(['mean', 'sum', 'count'])
print(grouped_agg)
```

```python
# merging and joining DataFrames 
# create smapale dataframes 
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1':[1,2,3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value1':[10,20,30]})
```

```python
df1
```

```python
df2
```

```python
# merge DataFrame on the 'key colums'

pd.merge(df1,df2,on="Key",how="inner") # hame on= mtlb kon sa column lena hai aur how = "inner" is liye lia hai taaki hame pura mile 
```

```python
# merging using outer join

pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="left") # left means df1 is more priotized 29
```

```python
pd.merge(df1,df2,on="Key",how="right") # right side df2 is given more priority
```
