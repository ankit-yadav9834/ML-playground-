```python
import seaborn as sns 
```

```python
# Basic  plotings with Seaborn 
tips = sns.load_dataset('tips')
tips
```

```python
# creating scatter plot 
import matplotlib.pyplot as plt

sns.scatterplot(x ='total_bill', y = 'tip', data = tips)
plt.title("Scatter plot of total Bil vs Tip")
plt.show()
```

```python
# Line plot

sns.lineplot(x = 'size', y = 'total_bill', data = tips)
plt.title("Line plot of  total bill by Size")
plt.show()
```

```python
# Categorical Variable .
# Bar
sns.barplot(x= 'day', y = 'total_bill', data = tips)
plt.title("Bar plot of total bill")
plt.show()
```

```python
# Boxen plot

sns.boxenplot(x = 'day', y= 'total_bill', data = tips)
plt.title("Box plot of the total bill")
```

```python
# Box plot 
sns.boxplot(x = 'day', y= 'total_bill', data = tips)
plt.title("Box plot of the total bill")
```

```python
# Violin plot

sns.violinplot(x = 'day', y= 'total_bill', data = tips)
```

```python
## Histogram 
sns.histplot(tips['total_bill'], bins = 10, kde= True)
```

```python
sns.histplot(tips['total_bill'], bins = 10, kde= False)
```

```python
sns.histplot(tips['total_bill'], kde= True)
```

```python
# KDe plot 

sns.kdeplot(tips['total_bill'], shade= True) # instead of shade we use fill=True
```

```python
#Pairplot

sns.pairplot(tips)
```

```python
# heat map
corr = tips[['total_bill', 'tip', 'size']].corr()
corr
```

```python
sns.heatmap(corr, annot= True, cmap= 'coolwarm')
```
