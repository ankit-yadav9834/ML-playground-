```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,3,40,50]

#creating plot
plt.plot(x,y)
plt.xlabel("x axix ")
plt.ylabel("y axix")
plt.title("xy graph")
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,3,40,25]

plt.plot(x,y,color='red',linestyle= '-.', marker = 'o', linewidth= 1)
plt.grid(True)
```
